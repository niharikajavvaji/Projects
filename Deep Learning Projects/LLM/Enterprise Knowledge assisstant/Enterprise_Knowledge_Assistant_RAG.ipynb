```python
def load_documents():
  pdf_directory="/content/data"
  docs=[]

  for doc in Path(pdf_directory).glob("*.pdf"):
      loader=PyMuPDFLoader(doc)
      docs.extend(loader.load())

  return docs
```

```python
def split_documents(docs):
  text_Splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
  texts=text_Splitter.split_documents(docs)

  return texts
```

```python
def create_vector_store():
  model_name = "sentence-transformers/all-mpnet-base-v2"
  model_kwargs = {"device": "cpu"}
  encode_kwargs = {"normalize_embeddings": False}
  hf = HuggingFaceEmbeddings(
    model_name=model_name,
    model_kwargs=model_kwargs,
    encode_kwargs=encode_kwargs,
  )

  return hf
```

```python
def get_retriever():
  retriever=chroma_db.as_retriever(search_type="similarity",search_kwargs={"k": 4})
  return retriever
```

```python
def retrieve_documents(question, retriever):
  retrieved_docs = retriever.invoke(question)
  return retrieved_docs
```

```python
def rerank_documents(question, retrieved_docs, reranker, top_k):
  pairs = [(question, doc.page_content) for doc in retrieved_docs]
  scores = reranker.predict(pairs)
  scored_docs = list(zip(retrieved_docs, scores))
  scored_docs = sorted(scored_docs, key=lambda x: x[1], reverse=True)
  top_docs = [doc for doc, score in scored_docs[:top_k]]
  return top_docs
```

```python
def build_prompt(context, question):
  # The 'system' message and 'template' are already defined globally from previous steps
  # This function now prepares the dictionary of inputs for the globally defined 'template' in the 'chain'
  return {
      "context": context,
      "question": question,
  }
```

```python
def invoke_llm(prompt,model,parser,query):
  # Assuming 'chain' is already defined globally from previous steps
  chain=template | model | parser

  result=chain.invoke({"context": formatted_context, "question": query})

  return result
```

```python
from pathlib import Path

def build_citations(top_docs):
    citations = []
    for doc in top_docs:
        file_name = Path(doc.metadata.get("file_path", "Unknown Source")).name
        page_num = str(doc.metadata.get("page", "Unknown Page"))
        citations.append(f"Source: {file_name}, Page: {page_num}")
    return "\n".join(citations)
```

```python
def generate_answer(llm_response, citations):
    return f"{llm_response}\n\n---\n\nCitations:\n{citations}"
```

## RAG Pipeline Full Implementation

All 11 methods for the RAG pipeline have been implemented. Now, let's put them all together to run the complete pipeline step-by-step.

```python
# 1. Load Documents
documents = load_documents()
print(f"Loaded {len(documents)} documents.")

# 2. Split Documents
chunks = split_documents(documents)
print(f"Split documents into {len(chunks)} chunks.")

# 3. Create Vector Store (get embedding model for Chroma)
embedding_model = create_vector_store()

# Assuming chroma_db is already initialized globally for simplicity here,
# or you would create it here if create_vector_store was modified to return it.
# For now, we'll assume it's still accessible globally from previous steps.
# If not, it needs to be initialized like: chroma_db = Chroma.from_documents(chunks, embedding_model)

# 4. Get Retriever
retriever = get_retriever()
print("Retriever created.")

# Define a query for demonstration
demo_query = "Explain Kubernetes.?"

# 5. Retrieve Documents
retrieved_docs = retrieve_documents(demo_query, retriever)
print(f"Retrieved {len(retrieved_docs)} documents for query: '{demo_query}'")

# 6. Rerank Documents (assuming 'reranker' is still globally defined)
top_k = 3 # Example top_k
top_docs = rerank_documents(demo_query, retrieved_docs, reranker, top_k)
print(f"Reranked documents, selected top {len(top_docs)}.")

# Format the context for the prompt
formatted_context = format_docs(top_docs)

# 7. Build Prompt
prompt = build_prompt(formatted_context, demo_query)
print("Prompt built.")

# Assuming 'chain' is defined globally from previous steps like: chain = template | model | parser
# If not, it needs to be initialized here with the model and parser

# 8. Invoke LLM
llm_response = invoke_llm(prompt,model,parser,demo_query)
print("LLM invoked, response received.")

# 9. Build Citations
citations = build_citations(top_docs)
print("Citations built.")

# 10. Generate Final Answer
final_answer = generate_answer(llm_response, citations)
print("Final Answer with citations:")
print(final_answer)
```
